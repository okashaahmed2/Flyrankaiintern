## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
from os import read
import pandas as pd

url = "https://raw.githubusercontent.com/okashaahmed2/Flyrankaiintern/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

total_pages = len(df)
print(f"Total number of pages: {total_pages}")

declining_pages=len(df[df['trend_direction'] == "down"])
print(f"Number of declining pages: {declining_pages}")

declining_pages_percentage = (declining_pages / total_pages) * 100
print(f"Percentage of declining pages: {declining_pages_percentage:.2f}%")

high_impact_declining = len(df[(df['trend_direction'] == 'down') & (df['impressions_90d'] >= 500)])
print(f"Number of high-impact declining pages: {high_impact_declining}")


```
